> ⚠️ **Audit note:** The app builds SQL by concatenating user input (`'insert … select \''+fn+'\',\''+rdc+'\''`), a SQL-injection pattern. A fruit name containing an apostrophe (e.g. *Dragon's Eye*) breaks the insert. Consider teaching parameter binding.

````
# SQL equivalent of the Streamlit wizard: the course's final app code (DWW_142, with the DWW_139 escape fix)
# is written to a stage and the app is created from it.
import io
from snowflake.snowpark.context import get_active_session
session = get_active_session()
session.sql(f"use role {ADMIN_ROLE}").collect()
APP_CODE = '# Import python packages\nimport streamlit as st\nfrom snowflake.snowpark.context import get_active_session\n\n# Write directly to the app\nst.title(":apple: Uncle Yer\'s Fruit Details :apple:")\nst.write(\n    """Enter fruit name and root depth code below.\n    """\n)\n\n# Get the current credentials\nsession = get_active_session()\n\n\nfn = st.text_input(\'Fruit Name:\')\nrdc = st.selectbox(\'Root Depth:\', (\'S\',\'M\',\'D\'))\n\nif st.button(\'Submit\'):\n    # st.write(\'Fruit Name entered is \' + fn)\n    # st.write(\'Root Depth Code chosen is \' + rdc)\n    sql_insert = \'insert into garden_plants.fruits.fruit_details select \\\'\'+fn+\'\\\',\\\'\'+rdc+\'\\\'\'\n    # st.write(sql_insert)\n    result = session.sql(sql_insert)\n    st.write(result)\n'
session.sql("create stage if not exists garden_plants.fruits.uncle_yers_app_stage").collect()
session.file.put_stream(io.BytesIO(APP_CODE.encode()), "@garden_plants.fruits.uncle_yers_app_stage/streamlit_app.py",
                        auto_compress=False, overwrite=True)
session.sql(f"""create or replace streamlit garden_plants.fruits.uncle_yers_root_depth_entry_form
  from '@garden_plants.fruits.uncle_yers_app_stage'
  main_file = 'streamlit_app.py'
  query_warehouse = {WAREHOUSE}
  title = 'Uncle Yers Root Depth Entry Form'""").collect()
print(APP_CODE)
````

> ⚠️ **Audit note:** Additions not in the course: the stage `GARDEN_PLANTS.FRUITS.UNCLE_YERS_APP_STAGE` holds the app file, and the app object name is chosen here (the UI generates one).

## Page: 🥋 Writing Our Entries to the Database

> **Course:** 🎯 Check your FRUIT_DETAILS table to see if you added a row. Once you have confirmed that your app works, add two more rows (for a total of 3). You can use any fruit names with any root depth settings.

````
# A notebook cannot click the app's Submit button. This runs the exact statement the app builds,
# once per entry. Apricot/S matches DWW_142; the other two are chosen here.
from snowflake.snowpark.context import get_active_session
session = get_active_session()
for fn, rdc in [("Apricot", "S"), ("Blueberry", "S"), ("Cherry", "M")]:
    sql_insert = 'insert into garden_plants.fruits.fruit_details select \''+fn+'\',\''+rdc+'\''
    print(sql_insert, "->", session.sql(sql_insert).collect())
````
